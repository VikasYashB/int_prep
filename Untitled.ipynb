```python
def factorial(value):
    result = 1
    while value > 0:
        result  = result * value
        value = value - 1

    return result

factorial(5)
```

```python
import apache_beam as beam

def to_csv(row):
    key, value = row
    data1_rows = value.get('data1', [])
    data2_rows = value.get('data2', [])
    data1_str = ','.join(data1_rows[0]) if data1_rows else ''
    data2_str = ','.join(data2_rows[0]) if data2_rows else ''
    return f"{key},{data1_str},{data2_str}"


p = beam.Pipeline()

data1 = (
		p
		| "read data file 1" >> beam.io.ReadFromText(r"C:\Users\Prasad\Documents\python_tests\data.csv",skip_header_lines=0)
		| "split columns" >> beam.Map(lambda row: row.split(','))
        # | "show data" >> beam.Map(print)
		| "create kv" >> beam.Map(lambda row: (row[0], row))

		)
		
data2 = (
		p
		| "read data file 2" >> beam.io.ReadFromText(r"C:\Users\Prasad\Documents\python_tests\data2.csv",skip_header_lines=0)
		| "split columns 2" >> beam.Map(lambda row: row.split(','))
        # | "show data" >> beam.Map(print)
		| "create kv 2" >> beam.Map(lambda row: (row[1], row))
		)
		
joined = (
		{"data1" : data1, "data2": data2}
		| "join " >> beam.CoGroupByKey()
		| "show joined data" >> beam.Map(print)
		)

csv_pcoll = (
    joined
    | "to csv" >> beam.Map(to_csv)
    | "write csv" >> beam.io.WriteToText("joined_output.csv", file_name_suffix=".csv", shard_name_template="")
)

p.run()
```
