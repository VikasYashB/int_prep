## find total displacement in below two arrays


```python
## find total displacement in below two arrays

Array1 = [1,2,4,8] 
Array2 = [1,4,2,8]
disp = 0
for i in range(len(Array1)):
    a1_el = Array1[i]
    a2_el_index = Array2.index(a1_el)
    index_dis = abs(a2_el_index - i)
    disp = disp + index_dis
disp
```

## keep rotating string1 and check if it matches string2


```python
## keep rotating string1 and check if it matches string2
string_1 = 'abcd' 
string_2 = 'cdab'

def is_rotation(s1, s2):
    if len(s1) != len(s2):
        return False
    for i in range(len(s1)):
        # print(s1[i:])
        # print(s1[:i])
        rotated = s1[i:] + s1[:i]
        print(rotated)
        if rotated == s2:
            return True
    return False

is_rotation(string_1, string_2)
```

# what are decorators, how to use them?

```python
from functools import wraps
import time

# What is a decorator?
# A decorator is a callable that takes a function (or class) and returns a new function (or class),
# usually to add behavior before/after the original callable runs.
# Use with the @decorator syntax above the function definition.


# Simple decorator that logs call info
def logger(func):
	@wraps(func)
	def wrapper(*args, **kwargs):
		print(f"Calling {func.__name__} with args={args}, kwargs={kwargs}")
		result = func(*args, **kwargs)
		print(f"{func.__name__} returned {result!r}")
		return result
	return wrapper

@logger
def add(x, y):
	return x + y

# Parameterized decorator (a decorator factory)
def repeat(times):
	def decorator(func):
		@wraps(func)
		def wrapper(*args, **kwargs):
			result = None
			for _ in range(times):
				result = func(*args, **kwargs)
			return result
		return wrapper
	return decorator

@repeat(3)
def greet(name):
	print(f"Hello, {name}!")

# Example showing preservation of metadata via functools.wraps
@logger
def slow_square(n):
	"""Return n squared after a short delay."""
	time.sleep(0.01)
	return n * n

# Usage examples
if __name__ == "__main__":
	print(add(2, 3))        # logger prints before/after, then prints 5
	greet("Alice")          # prints greeting 3 times
	print(slow_square.__name__, "->", slow_square.__doc__)
	print(slow_square(4))
```

### Python code to generate fibonacci series

```python
def fibonacci(n):
    s = [0,1]
    for i in range(n-2):
        s.append(s[-1] + s[-2])

    return s

s = fibonacci(12)
print(s)
print(len(s))
```

```python
numbers = [1, 2, 3, 4]
# Output: ['Odd', 'Even', 'Odd', 'Even']
# Using List Comprehension, 
# Using a Standard for Loop, Using map() with a lambda function

output = ['Odd' if x % 2 == 1 else 'Even' for x in numbers]
print('list comprehension ', output)

res = lambda x: 'Odd' if x%2==1 else 'Even'
output = []
for x in numbers:
	output.append(res(x))

print('lambda for loop ', output)

output = []
for x in numbers:
	if x%2 ==1:
		output.append('Odd')
	else:
		output.append('Even')
		
print('for loop ', output)
```

Element counts: {'apple': 3, 'banana': 2, 'orange': 1}

```python
elements = ['apple', 'banana', 'apple', 'orange', 'banana', 'apple']

ele_count = {}

for el in elements:
    if el in ele_count:
        ele_count[el] += 1
    else:
        ele_count[el] = 1

duplicates = {}

for k, v in ele_count.items():
    if v > 1:
        duplicates[k] = v
ele_count
```

#### remove duplicated from below list
records = [{"id":1},{"id":2},{"id":1}]


```python
records = [{"id":1},{"id":2},{"id":1}]

seen = []
result = []
for el in records:
	v = el['id']
	if v in seen:
		continue
	else:
		result.append(el)
		seen.append(v)

result
```

```python
seen = set()
result = [el for el in records if not (el["id"] in seen or seen.add(el["id"]))]
result
```
