## keep rotating string1 and check if it matches string2


```python
## keep rotating string1 and check if it matches string2
string_1 = 'abcd' 
string_2 = 'cdab'

def is_rotation(s1, s2):
    if len(s1) != len(s2):
        return False
    for i in range(len(s1)):
        # print(s1[i:])
        # print(s1[:i])
        rotated = s1[i:] + s1[:i]
        # print(rotated)
        if rotated == s2:
            return True
    return False

is_rotation(string_1, string_2)
```

# write python code to read from csv and write it to sql table.

```python
import os
from sqlalchemy import create_engine
import pandas as pd

csv_path = r'C:\Users\Prasad\Documents\python_tests\data\cust_transactions.csv'
df = pd.read_csv(csv_path)

db_url = r"sqlite:///C:/Users/Prasad/Documents/python_tests/my_database.db"
engine = create_engine(db_url)

table_name = 'cust_transactions'

try:
    df.to_sql(
        table_name,
        con=engine,
        if_exists='append',   # or 'replace'
        index=False,
        chunksize=1000
    )
    print(f"Wrote {len(df)} rows to table '{table_name}' on {db_url}")
finally:
    engine.dispose()
```

## Write python function to find number with most occurrence
input [1,2,2]
op : 2

input : [1,2,3]
op : 3

```python
def count_occ(ar):
    return max(set(ar), key=ar.count)

print(count_occ([1,2,2]))
print(count_occ([1,2,3]))
print(count_occ([1,2,3,4,4,4,4]))
```
