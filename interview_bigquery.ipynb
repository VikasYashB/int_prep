1. What is BigQuery and how does it differ from traditional databases?


```python
Detailed Answer:
BigQuery is a serverless, highly scalable, and cost-effective enterprise data warehouse designed for 
analytics on large datasets. Unlike traditional RDBMS:

Serverless: No infrastructure provisioning or management.
Columnar Storage: Optimized for analytical queries, reducing scan costs.
Distributed Query Execution: Uses Dremel engine for parallel processing.
Separation of Storage & Compute: Enables independent scaling and cost optimization.
Pay-as-you-go: You pay for the data processed, not for idle resources.

Example Use Case:
If you have petabytes of clickstream data, BigQuery can run complex aggregations in seconds 
without managing clusters like in Hadoop or Spark.
```

2. Explain BigQuery architecture.


```python
Detailed Answer:
BigQuery consists of:

Storage Layer: Data stored in Colossus, Google's distributed file system, in a columnar format.
Compute Layer: Queries executed using Dremel, which breaks queries into trees and processes them 
in parallel.
Service Layer: Handles API requests, authentication, and job scheduling.

Key Features:
Decoupled Storage & Compute: Scale independently.
Global Availability: Data replicated across regions for durability.
Streaming & Batch Support: Real-time ingestion and scheduled loads.
```

### 4. What are BigQuery partitions and clustering?


```python
Detailed Answer:

Partitioning:
Splits a table into segments based on a column (e.g., DATE).
Reduces scanned data for time-based queries.
Types: Time-based, Integer range, Ingestion time.

Clustering:
Organizes data within partitions based on columns.
Improves query performance for filters and aggregations.
```

### how to create this table in bigquery
Table Name: TRANSACTION

Column Name:  Transaction_date, customer_name, customer_city, Trans_no, Amount_Spent


```python
CREATE TABLE IF NOT EXISTS `projectID.datasetID.TRANSACTION` (
  Transaction_date DATETIME,
  customer_name STRING,
  customer_city STRING,
  Trans_no STRING,
  Amount_Spent INT64
)
PARTITION BY DATE(Transaction_date)
CLUSTER BY customer_city;
```

####    How do you query data in BigQuery?

```python
1- Using bigquery console UI 
2- in CLI using bq command: bq query "select columns from projectID.datasetID.table"
3- BigQuery API (REST API) 
    Use payload as below and create API request.
    {
        "query": "SELECT * FROM `project.dataset.table` LIMIT 10",
        "useLegacySql": false
        }
4- Using python script:
from google.cloud import bigquery

client = bigquery.Client()
query = "SELECT * FROM `project.dataset.table` LIMIT 10"
results = client.query(query)

for row in results:
    print(row)

5- In Airflow:
BigQueryInsertJobOperator(
    task_id="run_query",
    configuration={
        "query": {
            "query": "SELECT COUNT(*) FROM `proj.ds.table`",
            "useLegacySql": False,
        }
    }
)

```
